```python
# ==========================================
# TASK 4: IMPROVED eLCS 
# ==========================================

import numpy as np
import time
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
from skeLCS import eLCS

print("Loading preprocessed datasets from previous task...")
# These commands read the exact files saved by Task 3
X_train_resampled = np.load('X_train_resampled.npy')
y_train_resampled = np.load('y_train_resampled.npy')
X_test_clean = np.load('X_test_clean.npy')
y_test = np.load('y_test.npy')

print(f"Data Loaded Successfully! Training size: {X_train_resampled.shape}")

# 1. Initialize the eLCS model
print("\nRunning Improved eLCS on Clean, Balanced Data...")
improved_model = eLCS(
    learning_iterations=5000,
    N=500,
    p_spec=0.5
)

# 2. Train the model using the loaded training data
start_time_imp = time.time()
improved_model.fit(X_train_resampled, y_train_resampled)
training_time_imp = time.time() - start_time_imp
print(f"Improved Training Time: {training_time_imp:.2f} seconds")

# 3. Predict on the clean, unseen Test Data
y_pred_imp = improved_model.predict(X_test_clean)

# 4. Calculate performance metrics
imp_acc = accuracy_score(y_test, y_pred_imp)
imp_prec = precision_score(y_test, y_pred_imp, average='weighted')
imp_rec = recall_score(y_test, y_pred_imp, average='weighted')
imp_f1 = f1_score(y_test, y_pred_imp, average='weighted')

print("\n=== IMPROVED eLCS RESULTS ===")
print(f"Accuracy:  {imp_acc:.4f}")
print(f"F1-Score:  {imp_f1:.4f}")

# ==========================================
# PRINTING THE COMPARISON REPORT
# ==========================================
# NOTE: Update these numbers with Task 2 Baseline scores
base_acc, base_prec, base_rec, base_f1 = 0.7231, 0.6845, 0.7231, 0.6912  

print("\n" + "="*50)
print("       FINAL ASSIGNMENT COMPARISON MATRIX")
print("="*50)
print(f"{'Metric':<15} | {'Baseline eLCS':<15} | {'Improved eLCS':<15} | {'Delta (Change)':<12}")
print("-" * 65)
print(f"{'Accuracy':<15} | {base_acc:<15.4f} | {imp_acc:<15.4f} | {imp_acc - base_acc:<+12.4f}")
print(f"{'Precision':<15} | {base_prec:<15.4f} | {imp_prec:<15.4f} | {imp_prec - base_prec:<+12.4f}")
print(f"{'Recall':<15} | {base_rec:<15.4f} | {imp_rec:<15.4f} | {imp_rec - base_rec:<+12.4f}")
print(f"{'F1-Score':<15} | {base_f1:<15.4f} | {imp_f1:<15.4f} | {imp_f1 - base_f1:<+12.4f}")
print("="*50)
```

```python
# =========================================================
# HYPERPARAMETER TUNING FOR IMPROVED eLCS
# =========================================================

import time
from skeLCS import eLCS
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score

print("Initializing Tuned eLCS Model...")
# We dramatically increase N (population) and learning_iterations (evolutionary cycles)
tuned_elcs_params = {
    'learning_iterations': 20000,  # Scaled up from 5,000 (Gives rules more time to evolve)
    'N': 2000,                     # Scaled up from 500   (Allows the model to hold more complex rules)
    'p_spec': 0.5
}

tuned_model = eLCS(**tuned_elcs_params)

print("Training Tuned eLCS on Balanced Dataset (This may take a few minutes)...")
start_time = time.time()
tuned_model.fit(X_train_resampled, y_train_resampled)
print(f"Tuned Training Time: {time.time() - start_time:.2f} seconds")

# Evaluate the results
y_pred_tuned = tuned_model.predict(X_test_clean)

print("\n=== TUNED IMPROVED eLCS RESULTS ===")
print(f"Accuracy:  {accuracy_score(y_test, y_pred_tuned):.4f}")
print(f"Precision: {precision_score(y_test, y_pred_tuned, average='weighted'):.4f}")
print(f"Recall:    {recall_score(y_test, y_pred_tuned, average='weighted'):.4f}")
print(f"F1-Score:  {f1_score(y_test, y_pred_tuned, average='weighted'):.4f}")
```
